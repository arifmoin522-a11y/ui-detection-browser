# Export web-form-ui-field-detection to ONNX

This notebook exports the [foduucom/web-form-ui-field-detection](https://huggingface.co/foduucom/web-form-ui-field-detection) YOLOv8s model to ONNX format for browser inference.

```python
!pip install ultralyticsplus==0.0.28 ultralytics==8.0.43
```

```python
from ultralyticsplus import YOLO

model = YOLO('foduucom/web-form-ui-field-detection')

model.overrides['conf'] = 0.25
model.overrides['iou'] = 0.45
model.overrides['agnostic_nms'] = False
model.overrides['max_det'] = 1000
```

```python
model.export(format='onnx', imgsz=640, simplify=True)
print('Export complete: best.onnx')
```

```python
import os
os.rename('best.onnx', '../model/ui-detection.onnx')
print(f'Model size: {os.path.getsize("../model/ui-detection.onnx") / 1024 / 1024:.1f} MB')
```
